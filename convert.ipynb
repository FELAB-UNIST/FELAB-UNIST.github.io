```python
"""
BibTeX to JSON Converter
- Jupyter notebook에서 사용
- Google Scholar에서 복사한 BibTeX를 JSON 필드에 넣을 수 있는 형태로 변환
"""

def bibtex_to_json(bibtex: str) -> str:
    """
    여러 줄 BibTeX를 JSON 문자열로 변환
    
    Usage:
        bibtex = '''@article{kim2025guruagents,
          title={GuruAgents: Emulating Wise Investors},
          author={Kim, Yejin and Lee, Youngbin},
          year={2025}
        }'''
        
        result = bibtex_to_json(bibtex)
        print(result)  # 복사해서 JSON에 붙여넣기
    """
    # 앞뒤 공백 제거하고 줄바꿈을 \n으로 변환
    escaped = bibtex.strip().replace('\n', '\\n')
    return escaped


def print_json_bibtex(bibtex: str):
    """변환 결과를 보기 좋게 출력"""
    result = bibtex_to_json(bibtex)
    print("=" * 60)
    print("JSON에 복사할 문자열:")
    print("=" * 60)
    print(f'"{result}"')
    print("=" * 60)


# 클립보드 지원 (pyperclip 설치된 경우)
def copy_json_bibtex(bibtex: str):
    """변환 후 클립보드에 복사"""
    try:
        import pyperclip
        result = bibtex_to_json(bibtex)
        pyperclip.copy(f'"{result}"')
        print("✅ 클립보드에 복사되었습니다!")
        print(f"\n미리보기: {result[:80]}...")
    except ImportError:
        print("pyperclip이 설치되지 않았습니다. pip install pyperclip")
        print_json_bibtex(bibtex)
```

```python
if __name__ == "__main__":
    # 예시 BibTeX
    sample = """@article{kim2020personalized,
  title={Personalized goal-based investing via multi-stage stochastic goal programming},
  author={Kim, Woo Chang and Kwon, Do-Gyun and Lee, Yongjae and Kim, Jang Ho and Lin, Changle},
  journal={Quantitative Finance},
  volume={20},
  number={3},
  pages={515--526},
  year={2020},
  publisher={Taylor \& Francis}
}"""
    
    print_json_bibtex(sample)
```
